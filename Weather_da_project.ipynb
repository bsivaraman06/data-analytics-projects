```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read Dataset
df = pd.read_csv("Weather History Dataset.csv")
```

```python
# Check First 5 Rows
print("\n--- First 5 Rows ---")
print(df.head())
```

```python
# Check Last 5 Rows
print("\n--- Last 5 Rows ---")
print(df.tail())
```

```python
# Check Random 5 Rows
print("\n--- Random 5 Rows ---")
print(df.sample(5))
```

```python
# 5. Check Number of Rows and Columns
print("\n--- Dataset Shape ---")
print(df.shape)
```

```python
# Check Column Names
print("\n--- Column Names ---")
print(df.columns.tolist())
```

```python
# Check Data Types and Non-Null Counts
print("\n--- Dataset Information ---")
df.info()
```

```python
# Statistical Summary
print("\n--- Statistical Summary ---")
print(df.describe())
```

```python
# Check Missing Values
print("\n--- Missing Values ---")
print(df.isnull().sum())
```

```python
# Drop Duplicates

print("Duplicate Rows Before Dropping:")
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicate Rows After Dropping:")
print(df.duplicated().sum())

print("Dataset Shape After Dropping Duplicates:")
print(df.shape)
```

```python
# Filter temperature above 20°C

print("Temperature Above 20°C:")
print(df[df["Temperature (C)"] > 20])
```

```python
# Humidity filter

print("Humidity Above 80%:")
print(df[df["Humidity"] > 0.80])
```

```python
# Specific weather condition filter

print("Rain Weather Records:")
print(df[df["Summary"] == "Rain"])
```

```python
# Multiple conditions filter

print("Temperature Above 20°C and Humidity Above 80%:")
print(df[(df["Temperature (C)"] > 20) & (df["Humidity"] > 0.80)])
```

```python
# Filter using isin()

print("Rain or Snow Records:")
print(df[df["Precip Type"].isin(["rain", "snow"])])
```

```python
# Print filtered row count

filtered_df = df[df["Temperature (C)"] > 20]
print("Filtered Data:")
print(filtered_df)
print("Filtered Rows Count:")
print(filtered_df.shape[0])
```

```python
# groupby() — Average temperature by weather

print("Average Temperature by Weather:")
print(df.groupby("Summary")["Temperature (C)"].mean())
```

```python
# groupby() — Average humidity by weather

print("Average Humidity by Weather:")
print(df.groupby("Summary")["Humidity"].mean())
```

```python
# groupby() — Multiple statistical values

print("Temperature Statistics by Weather:")
print(df.groupby("Summary")["Temperature (C)"].agg(["mean", "min", "max", "count"]))
```

```python
# value_counts() — Weather condition counts

print("Weather Condition Counts:")
print(df["Summary"].value_counts())
```

```python
# value_counts() — Precipitation types

print("Precipitation Type Counts:")
print(df["Precip Type"].value_counts(dropna=False))
```

```python
# groupby() — Average wind speed by precipitation type

print("Average Wind Speed by Precipitation Type:")
print(df.groupby("Precip Type")["Wind Speed (km/h)"].mean())
```

```python
# groupby() — Temperature and humidity together

print("Weather-wise Temperature and Humidity:")
print(df.groupby("Summary")[["Temperature (C)", "Humidity"]].mean())
```

```python
# Temperature Distribution

plt.figure(figsize=(8, 6))

sns.histplot(
    data=df,
    x="Temperature (C)",
    bins=12,
    kde=True,
    edgecolor="white",
    linewidth=1.2
)

plt.title("Temperature Distribution", fontsize=15)
plt.xlabel("Temperature (C)")
plt.ylabel("Frequency")

plt.grid(False)

plt.tight_layout()

plt.show()
```

```python
# Temperature Boxplot

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x="Temperature (C)",
    color="coral"
)

plt.title("Temperature Boxplot", fontsize=20)
plt.xlabel("Temperature (C)")
plt.ylabel("Distribution")

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Precipitation Type Count

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x="Precip Type",
    color="orchid"
)

plt.title("Precipitation Type Count", fontsize=15)
plt.xlabel("Precipitation Type")
plt.ylabel("Count")

plt.tight_layout()

plt.show()
```

```python
# Temperature vs Apparent Temperature

plt.figure(figsize=(6, 6))

sns.scatterplot(
    data=df.head(10),
    x="Temperature (C)",
    y="Apparent Temperature (C)",
    alpha=0.4,
    color="red"
)

plt.title("Actual vs Apparent Temperature", fontsize=15)
plt.xlabel("Temperature (C)")
plt.ylabel("Apparent Temperature (C)")

plt.grid(False)


plt.tight_layout()
plt.show()
```

```python
# Temperature by Precipitation Type

plt.figure(figsize=(8, 6))

sns.violinplot(
    data=df,
    x="Precip Type",
    y="Temperature (C)",
    color="gold"
)

plt.title("Temperature by Precipitation Type", fontsize=15)
plt.xlabel("Precipitation Type")
plt.ylabel("Temperature (C)")

plt.grid(True)


plt.tight_layout()
plt.show()
```

```python
# Humidity Distribution

plt.figure(figsize=(8, 6))

sns.kdeplot(
    data=df,
    x="Humidity",
    fill=True
)

plt.title("Humidity Distribution", fontsize=15)
plt.xlabel("Humidity")
plt.ylabel("Density")

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Weather Correlation Heatmap

plt.figure(figsize=(12, 8))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    annot_kws={"size":14,"weight":"bold"}
)

plt.title("Weather Variables Correlation", fontsize=12)
plt.xlabel("Weather Variables")
plt.ylabel("Weather Variables")

plt.tight_layout()
plt.show()
```

```python
# Pie Chart - Precipitation Type Distribution

plt.figure(figsize=(8, 8))

precip_counts = df["Precip Type"].value_counts()

plt.pie(
    precip_counts,
    labels=precip_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Precipitation Type Distribution", fontsize=15)

plt.tight_layout()
plt.show()
```

```python
# 4 Axes Colourful Weather Dashboard

plt.figure(figsize=(16, 10))

fig, axes = plt.subplots(2, 2, figsize=(24, 18))

fig.suptitle(
    "WEATHER DATA ANALYSIS DASHBOARD",
    fontsize=20,
    fontweight="bold"
)


# 1. Histogram - Temperature Distribution (First 5 Rows)

sns.histplot(
    data=df.head(5),
    x="Temperature (C)",
    bins=5,
    kde=False,
    color="orange",
    ax=axes[0, 0]
)

axes[0, 0].set_title("Temperature Distribution")
axes[0, 0].set_xlabel("Temperature (C)")
axes[0, 0].set_ylabel("Frequency")
axes[0, 0].grid(True)


# 2. Bar Chart - Humidity (First 5 Rows)

sns.barplot(
    data=df.head(5),
    x=df.head(5).index,
    y="Humidity",
    hue=df.head(5).index,
    palette="viridis",
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Humidity of First 5 Records")
axes[0, 1].set_xlabel("Record Index")
axes[0, 1].set_ylabel("Humidity")
axes[0, 1].grid(True)


# 3. Scatter Plot - Temperature vs Wind Speed

sns.scatterplot(
    data=df.head(10),
    x="Temperature (C)",
    y="Wind Speed (km/h)",
    hue="Summary",
    palette="Set1",
    s=150,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Temperature vs Wind Speed")
axes[1, 0].set_xlabel("Temperature (C)")
axes[1, 0].set_ylabel("Wind Speed (km/h)")
axes[1, 0].grid(True)



# 4. Pie Chart - Precipitation Type

precip_counts = df.head(5)["Precip Type"].value_counts()

axes[1, 1].pie(
    precip_counts,
    labels=precip_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2", len(precip_counts))
)

axes[1, 1].set_title("Precipitation Type Distribution")
# Dashboard title - dark and bold
fig.suptitle(
    "WEATHER DATA ANALYSIS DASHBOARD",
    fontsize=24,
    fontweight="bold",
    color="black"
)

# All chart titles, labels and ticks dark and bold
for ax in fig.axes:

    # Chart titles
    ax.title.set_fontsize(16)
    ax.title.set_fontweight("bold")
    ax.title.set_color("black")

    # X-axis and Y-axis labels
    ax.xaxis.label.set_fontsize(13)
    ax.xaxis.label.set_fontweight("bold")
    ax.xaxis.label.set_color("black")

    ax.yaxis.label.set_fontsize(13)
    ax.yaxis.label.set_fontweight("bold")
    ax.yaxis.label.set_color("black")

    # X-axis and Y-axis tick values
    ax.tick_params(
        axis="both",
        colors="black",
        labelsize=11,
        width=1.5
    )

    for label in ax.get_xticklabels():
        label.set_color("black")
        label.set_fontweight("bold")
        label.set_fontsize(11)

    for label in ax.get_yticklabels():
        label.set_color("black")
        label.set_fontweight("bold")
        label.set_fontsize(11)

    # Dark axis borders
    for spine in ax.spines.values():
        spine.set_color("black")
        spine.set_linewidth(1.5)

# Dark and bold text inside pie charts
for ax in fig.axes:
    for text_item in ax.texts:
        text_item.set_color("black")
        text_item.set_fontweight("bold")
        text_item.set_fontsize(11)





# Final Layout

plt.tight_layout()

plt.show()
```

```python
# 6 Axes Colourful Weather Dashboard

fig, axes = plt.subplots(
    3, 2,
    figsize=(22, 18),
    dpi=120
)

fig.suptitle(
    "WEATHER DATA ANALYSIS DASHBOARD",
    fontsize=26,
    fontweight="bold",
    color="darkblue",
    y=0.99
)

# Use First 5 Rows
data = df.head(5)


# 1. Histogram - Temperature Distribution

sns.histplot(
    data=data,
    x="Temperature (C)",
    bins=5,
    color="coral",
    ax=axes[0, 0]
)

axes[0, 0].set_title("Temperature Distribution", fontsize=15)
axes[0, 0].set_xlabel("Temperature (C)")
axes[0, 0].set_ylabel("Frequency")
axes[0, 0].grid(True, alpha=0.3)


# 2. Barplot - Humidity

sns.barplot(
    data=data,
    x=data.index,
    y="Humidity",
    hue=data.index,
    palette="viridis",
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Humidity of First 5 Records", fontsize=15)
axes[0, 1].set_xlabel("Record Index")
axes[0, 1].set_ylabel("Humidity")
axes[0, 1].grid(True, alpha=0.3)


# 3. Scatterplot - Temperature vs Wind Speed

sns.scatterplot(
    data=data,
    x="Temperature (C)",
    y="Wind Speed (km/h)",
    hue="Summary",
    palette="Set1",
    s=150,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Temperature vs Wind Speed", fontsize=15)
axes[1, 0].set_xlabel("Temperature (C)")
axes[1, 0].set_ylabel("Wind Speed (km/h)")
axes[1, 0].grid(True, alpha=0.3)


# 4. Boxplot - Temperature

sns.boxplot(
    data=data,
    y="Temperature (C)",
    color="mediumpurple",
    ax=axes[1, 1]
)

axes[1, 1].set_title("Temperature Boxplot", fontsize=15)
axes[1, 1].set_xlabel("First 5 Records")
axes[1, 1].set_ylabel("Temperature (C)")
axes[1, 1].grid(True, alpha=0.3)


# 5. Lineplot - Temperature vs Humidity

sns.lineplot(
    data=data,
    x=data.index,
    y="Temperature (C)",
    marker="o",
    color="green",
    linewidth=2.5,
    ax=axes[2, 0]
)

axes[2, 0].set_title("Temperature Trend", fontsize=15)
axes[2, 0].set_xlabel("Record Index")
axes[2, 0].set_ylabel("Temperature (C)")
axes[2, 0].grid(True, alpha=0.3)


# 6. Pie Chart - Precipitation Type

precip_counts = data["Precip Type"].value_counts()

axes[2, 1].pie(
    precip_counts,
    labels=precip_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2", len(precip_counts)),
    textprops={"fontsize": 12}
)

axes[2, 1].set_title("Precipitation Type Distribution", fontsize=15)
# Dashboard title - dark and bold
fig.suptitle(
    "WEATHER DATA ANALYSIS DASHBOARD",
    fontsize=24,
    fontweight="bold",
    color="black"
)

# All chart titles, labels and ticks dark and bold
for ax in fig.axes:

    # Chart titles
    ax.title.set_fontsize(16)
    ax.title.set_fontweight("bold")
    ax.title.set_color("black")

    # X-axis and Y-axis labels
    ax.xaxis.label.set_fontsize(13)
    ax.xaxis.label.set_fontweight("bold")
    ax.xaxis.label.set_color("black")

    ax.yaxis.label.set_fontsize(13)
    ax.yaxis.label.set_fontweight("bold")
    ax.yaxis.label.set_color("black")

    # X-axis and Y-axis tick values
    ax.tick_params(
        axis="both",
        colors="black",
        labelsize=11,
        width=1.5
    )

    for label in ax.get_xticklabels():
        label.set_color("black")
        label.set_fontweight("bold")
        label.set_fontsize(11)

    for label in ax.get_yticklabels():
        label.set_color("black")
        label.set_fontweight("bold")
        label.set_fontsize(11)

    # Dark axis borders
    for spine in ax.spines.values():
        spine.set_color("black")
        spine.set_linewidth(1.5)

# Dark and bold text inside pie charts
for ax in fig.axes:
    for text_item in ax.texts:
        text_item.set_color("black")
        text_item.set_fontweight("bold")
        text_item.set_fontsize(11)



# Final Dashboard Layout

plt.tight_layout(rect=[0, 0, 1, 0.96], pad=3.0)

plt.show()
```
